```python
# Task 3

# Problem Statement 1: Employee Performance Bonus Eligibility


# Employee performance scores
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Finding the highest score
highest_score = max(employees.values())

# Finding all employees with the highest score (handle ties)
top_performers = []

for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)

# To display result
print("Top Performers Eligible for Bonus:", 
      ", ".join(top_performers), 
      f"(Score: {highest_score})")
```

```python
# Problem Statement 2: Search Query Keyword Analysis

import string

# Given query
query = "Buy mobile phone buy phone online"

# Convert to lowercase
query = query.lower()

# To Remove punctuation
for char in string.punctuation:
    query = query.replace(char, "")

# Split query into words
words = query.split()

# Count frequency
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Displaying only keywords searched more than once
repeated_keywords = {word: count for word, count in word_count.items() if count > 1}

print(repeated_keywords)
```

```python
# Problem Statement 3: Sensor Data Validation

# Sensor readings given
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

# Creating list to store valid readings (hour_index, reading_value)
valid_readings = []

# Initialising loop through readings using enumerate to get index and value
for hour, reading in enumerate(sensor_readings):
    if reading % 2 == 0:  # Check if even (valid)
        valid_readings.append((hour, reading))

# Displaying result
print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

```python
# Problem Statement 4: Email Domain Usage Analysis

# Email list
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

# Creating dictionary to store domain counts
domain_count = {}

# Extracting domains and count
for email in emails:
    domain = email.split("@")[1]  # Extract domain
    
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

# Calculate and display percentage usage
total_emails = len(emails)

for domain, count in domain_count.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {percentage:.0f}%")
```

```python
# Problem Statement 5: Sales Spike Detection

# Sales data list
sales = [1200, 1500, 900, 2200, 1400, 3000]

# Calculating average sales
average_sales = sum(sales) / len(sales)

# Calculating spike threshold (30% above average)
threshold = average_sales * 1.3

# Detect spike days
for index, value in enumerate(sales):
    if value > threshold:
        print(f"Day {index + 1}: {value}")
```

```python
# Problem Statement 6: Duplicate User ID Detection

# User IDs list
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

# Creating dictionary to store frequency
user_count = {}

# To count occurrences
for user in user_ids:
    if user in user_count:
        user_count[user] += 1
    else:
        user_count[user] = 1

# Displaying duplicates only
for user, count in user_count.items():
    if count > 1:
        print(f"{user} → {count} times")
```
