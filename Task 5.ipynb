```python
#Problem 1: Social Media – Post Engagement Analyzer
def analyze_post_engagement(likes_list):
    total_likes = 0
    
    # Calculate total likes using loop
    for likes in likes_list:
        total_likes += likes
    
    # Determine post status
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    
    return total_likes, status


# Example likes data (from different platforms or time intervals)
likes = [200, 300, 150, 400]   # Total = 1050

# Function call
total, post_status = analyze_post_engagement(likes)

print(f"Total Likes: {total}")
print(f"Post Status: {post_status}")
```

```python
#Problem 2: Healthcare – Medicine Stock Alert System
def check_medicine_stock(stock):
    
    # Conditional logic
    if stock < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"
    
    return stock, status


# Example stock value
medicine_stock = 6

# Function call
stock_value, stock_status = check_medicine_stock(medicine_stock)

print(f"Medicine Stock: {stock_value}")
print(f"Status: {stock_status}")
```

```python
# Problem 3: Agriculture – Rainfall Adequacy Checker
def check_rainfall(rainfall_data, required_level):
    total_rainfall = 0
    
    # Calculate total rainfall using loop
    for rainfall in rainfall_data:
        total_rainfall += rainfall
    
    # Calculate average rainfall
    average_rainfall = total_rainfall / len(rainfall_data)
    
    # Determine rainfall status
    if average_rainfall >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"
    
    return average_rainfall, status


# Example seasonal rainfall data (in mm)
rainfall = [70, 75, 80, 63]   # Average = 72
required = 70

# Function call
avg, rainfall_status = check_rainfall(rainfall, required)

print(f"Average Rainfall: {int(avg)}")
print(f"Rainfall Status: {rainfall_status}")
```

```python
# Problem 4: Social Media – Duplicate Account Detection

def detect_duplicate_accounts(usernames):
    
    # Convert list to set
    unique_users = set(usernames)
    
    # If lengths differ → duplicates exist
    if len(unique_users) != len(usernames):
        result = "Yes"
    else:
        result = "No"
    
    return result


# Example usernames
accounts = ["user1", "user2", "user3", "user1"]

# Function call
duplicate_status = detect_duplicate_accounts(accounts)

print(f"Duplicate Accounts Found: {duplicate_status}")
```

```python
# Problem 5: Healthcare – Appointment Eligibility Checker
def check_appointment_eligibility(age):
    
    # Condition check
    if age >= 18:
        status = "Eligible"
    else:
        status = "Not Eligible"
    
    return age, status


# Example patient age
patient_age = 21

# Function call
age_value, eligibility_status = check_appointment_eligibility(patient_age)

print(f"Patient Age: {age_value}")
print(f"Eligibility Status: {eligibility_status}")
```

```python
# Problem 6: Agriculture – Premium Crop Price Filter
def filter_premium_crops(prices):
    premium_crops = []
    
    # Loop and filter prices
    for price in prices:
        if price > 2000:
            premium_crops.append(price)
    
    return premium_crops


# Example crop prices
crop_prices = [1500, 2500, 1800, 3200, 1900]

# Function call
premium_list = filter_premium_crops(crop_prices)

print(f"Premium Crops: {premium_list}")
```

```python
# Problem 7: System Monitoring – Application Health Checker
def check_application_health(error_count):
    
    # Conditional checks
    if error_count == 0:
        status = "Healthy"
    elif error_count <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"
    
    return error_count, status


# Example error count
errors = 7

# Function call
error_value, system_status = check_application_health(errors)

print(f"Error Count: {error_value}")
print(f"System Status: {system_status}")
```

```python
# Problem 8: Banking – Daily Transaction Limit Checker
def check_transaction_limit(amount):
    daily_limit = 50000
    
    # Conditional check
    if amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"
    
    return amount, status


# Example transaction amount
transaction_amount = 60000

# Function call
amount_value, transaction_status = check_transaction_limit(transaction_amount)

print(f"Transaction Amount: {amount_value}")
print(f"Transaction Status: {transaction_status}")
```

```python
# Problem 9: E-Learning – Student Attendance Eligibility System
def check_exam_eligibility(attendance_list):
    present_days = 0
    
    # Count present days using loop
    for day in attendance_list:
        if day == "P":
            present_days += 1
    
    total_days = len(attendance_list)
    
    # Calculate attendance percentage
    attendance_percentage = (present_days / total_days) * 100
    
    # Determine eligibility
    if attendance_percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"
    
    return attendance_percentage, status


# Example attendance record
attendance = ["P", "P", "A", "P", "P"]  # 4 out of 5 present → 80%

# Function call
percentage, eligibility = check_exam_eligibility(attendance)

print(f"Attendance Percentage: {percentage}")
print(f"Exam Eligibility: {eligibility}")
```

```python
# Problem 10: Smart Electricity Bill Analyzer
def calculate_electricity_bill(units):
    bill = 0
    
    # Slab calculation
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
    
    # Usage classification
    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"
    
    return bill, status


# Example units consumed
units_consumed = 250

# Function call
total_bill, usage_status = calculate_electricity_bill(units_consumed)

print(f"Units Consumed: {units_consumed}")
print(f"Total Bill: ₹{total_bill}")
print(f"Usage Status: {usage_status}")
```
