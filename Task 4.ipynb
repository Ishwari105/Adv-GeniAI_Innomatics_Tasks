```python
# Task 4

# Problem 1: Mobile Recharge Validation System

def validate_recharge():
    valid_plans = [199, 299, 399, 599]
    
    while True:
        try:
            amount = int(input("Enter recharge amount: ₹"))
            
            # Rule 1: Minimum amount check
            if amount < 50:
                print("Minimum recharge amount is ₹50. Please try again.\n")
                continue
            
            # Rule 2: Valid plan check
            if amount in valid_plans:
                print("Recharge Successful!")
                break
            else:
                print("Invalid plan selected.")
                print(f"Available plans: {valid_plans}\n")
                
        except ValueError:
            print("Please enter a valid numeric amount.\n")


# Run the function
validate_recharge()
```

```python
# Problem 2: Inventory Reorder Alert System

def check_inventory(products):
    for product, quantity in products.items():
        if quantity < 15:
            print(f"{product}: Reorder Alert (Stock: {quantity})")
        else:
            print(f"{product}: Stock OK (Stock: {quantity})")


# Take an inventory data
inventory = {
    "Rice": 25,
    "Wheat": 10,
    "Sugar": 18,
    "Oil": 8
}

# Calling the function
check_inventory(inventory)
```

```python
# Problem 3: Student Result Processing System

def process_result(marks):
    total = 0
    
    # Calculate total using loop
    for mark in marks:
        total += mark
    
    # To calculate average
    average = total / len(marks)
    
    # To determine result
    if average >= 50:
        result = "Pass"
    else:
        result = "Fail"
    
    return average, result


# Take sample marks
student_marks = [60, 45, 70, 55]

# Function call
average_marks, final_result = process_result(student_marks)

print(f"Average Marks: {average_marks}")
print(f"Result: {final_result}")
```

```python
# Problem 4: Cab Fare Estimator with Retry Option

def calculate_fare(distance, is_peak):
    base_fare = 50
    per_km_rate = 12
    
    # Calculating fare
    fare = base_fare + (distance * per_km_rate)
    
    # Applying peak hour charges (25% extra)
    if is_peak:
        fare += fare * 0.25
    
    return fare


# To retry system using while loop
while True:
    try:
        distance = float(input("Enter distance in km: "))
        peak_input = input("Is it peak hour? (yes/no): ").lower()
        
        if peak_input == "yes":
            peak = True
        elif peak_input == "no":
            peak = False
        else:
            print("Invalid peak hour input. Please enter yes or no.\n")
            continue
        
        total_fare = calculate_fare(distance, peak)
        
        print(f"Total Fare: ₹{total_fare:.2f}")
        
        retry = input("Do you want to calculate again? (yes/no): ").lower()
        if retry != "yes":
            print("Thank you for using Cab Fare Estimator!")
            break
            
    except ValueError:
        print("Please enter valid numeric distance.\n")
```

```python
# Problem 5: Employee Attendance Eligibility Checker

def check_attendance(attendance_list):
    present_count = 0
    
    # Counting present days using loop
    for status in attendance_list:
        if status == "P":
            present_count += 1
    
    total_days = len(attendance_list)
    
    # Calculating percentage
    attendance_percentage = (present_count / total_days) * 100
    
    # Determining eligibility
    if attendance_percentage >= 75:
        result = "Eligible"
    else:
        result = "Not Eligible"
    
    return attendance_percentage, result


# To sample attendance data
attendance = ["P", "P", "A", "P", "P", "A", "P", "P"]

# Calling function 
percentage, eligibility = check_attendance(attendance)

print(f"Attendance Percentage: {percentage:.2f}%")
print(f"Status: {eligibility}")
```

```python
# Problem 6: Password Strength Checker

def check_password_strength(password):
    has_digit = False
    has_special = False
    special_chars = "@#$"
    
    # Minimum length check
    if len(password) < 8:
        return "Weak Password (Minimum length should be 8)"
    
    # Loop through password characters
    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True
    
    # Final validation
    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password (Must include at least one digit and one special character @#$)"


# Taking an example 
user_password = input("Enter password: ")
result = check_password_strength(user_password)
print(result)
```
