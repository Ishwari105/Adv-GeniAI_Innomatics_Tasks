```python
# Problem Statement 1:Smart Parking Lot Management System

def manage_parking(capacity, vehicle_logs):
    current_parked = 0
    peak_parked = 0
    
    # Process logs using loop
    for log in vehicle_logs:
        if log == "IN":
            current_parked += 1
        elif log == "OUT" and current_parked > 0:
            current_parked -= 1
        
        # Track peak usage
        if current_parked > peak_parked:
            peak_parked = current_parked
    
    # Check parking status
    if current_parked > capacity:
        status = "Alert: Parking Full / Over Capacity"
    else:
        status = "Available"
    
    return current_parked, peak_parked, status


# Sample Input
parking_capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

# Function call
current, peak, parking_status = manage_parking(parking_capacity, vehicle_logs)

print(f"Currently Parked Vehicles: {current}")
print(f"Peak Parking Usage: {peak}")
print(f"Parking Status: {parking_status}")
```

```python
# Problem Statement 2:Online Food Delivery Time Estimator

def estimate_delivery_time(distance, traffic_level, weather):
    # Base time calculation
    base_time = distance * 5
    
    # Traffic delay
    if traffic_level == "Low":
        traffic_delay = 5
    elif traffic_level == "Medium":
        traffic_delay = 10
    elif traffic_level == "High":
        traffic_delay = 15
    else:
        traffic_delay = 0
    
    # Weather delay
    if weather == "Clear":
        weather_delay = 0
    elif weather == "Rainy":
        weather_delay = 10
    elif weather == "Stormy":
        weather_delay = 20
    else:
        weather_delay = 0
    
    # Final ETA
    total_time = base_time + traffic_delay + weather_delay
    
    return total_time


# Sample Input
distance = 8
traffic = "High"
weather_condition = "Rainy"

eta = estimate_delivery_time(distance, traffic, weather_condition)

print(f"Estimated Delivery Time: {eta} minutes")
```

```python
# Problem Statement 3:Movie Theatre Seat Occupancy Analyzer

def analyze_seat_occupancy(total_seats, booked_seats):
    
    # Count booked seats
    booked_count = len(booked_seats)
    
    # Calculate occupancy percentage
    occupancy_percentage = (booked_count / total_seats) * 100
    
    # Determine show status
    if occupancy_percentage == 100:
        status = "Housefull"
        suggestion = "Open Additional Show"
    elif occupancy_percentage >= 75:
        status = "Almost Full"
        suggestion = "Consider Opening Additional Show"
    else:
        status = "Seats Available"
        suggestion = "No Additional Show Needed"
    
    return occupancy_percentage, status, suggestion


# Sample Input
total_seats = 200
booked_seats = [1] * 150   # 150 seats booked

occupancy, show_status, suggestion = analyze_seat_occupancy(total_seats, booked_seats)

print(f"Occupancy: {occupancy:.0f}%")
print(f"Show Status: {show_status}")
print(f"Suggestion: {suggestion}")
```

```python
# Problem Statement 4:Cloud Server Load Classification System

def classify_server_load(cpu_readings):
    
    # Calculate total using loop
    total = 0
    for reading in cpu_readings:
        total += reading
    
    # Calculate average
    average_cpu = total / len(cpu_readings)
    
    # Classify server status
    if average_cpu < 50:
        status = "Normal"
    elif 50 <= average_cpu <= 80:
        status = "Warning"
    else:
        status = "Critical"
    
    return average_cpu, status


# Sample Input
cpu_data = [45, 60, 70, 85, 90]

avg_cpu, server_status = classify_server_load(cpu_data)

print(f"Average CPU Load: {avg_cpu:.0f}%")
print(f"Server Status: {server_status}")
```

```python
# Problem Statement 5:Smart Classroom Resource Usage Monitor

def monitor_resource_usage(resource_usage):
    
    overused_resources = []
    threshold = 8   # hours limit
    
    # Check each resource usage
    for resource, hours in resource_usage.items():
        if hours > threshold:
            overused_resources.append(resource)
    
    # Energy alert condition
    if len(overused_resources) > 0:
        alert = "Yes"
    else:
        alert = "No"
    
    return overused_resources, alert


# Sample Input
usage_data = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

overused, energy_alert = monitor_resource_usage(usage_data)

print("Overused Resources:", ", ".join(overused) if overused else "None")
print("Energy Alert:", energy_alert)
```

```python
# Problem Statement 6: Online Event Registration Capacity Controller

def manage_event_registrations(capacity, total_registrations):
    
    confirmed = 0
    waitlisted = 0
    
    # Simulate registration processing using loop
    for i in range(total_registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1
    
    # Registration status
    if confirmed >= capacity:
        status = "Closed"
    else:
        status = "Open"
    
    return confirmed, waitlisted, status


# Sample Input
event_capacity = 100
registrations = 105

confirmed_count, waitlisted_count, registration_status = manage_event_registrations(event_capacity, registrations)

print(f"Confirmed Registrations: {confirmed_count}")
print(f"Waitlisted Users: {waitlisted_count}")
print(f"Registration Status: {registration_status}")
```
