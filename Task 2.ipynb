```python
# Problem 1: Unique Words in a Sentence
```

```python
# Taking input from user
sentence = input("Enter sentence: ")

# Converting sentence to lowercase 
sentence = sentence.lower()

# Splitting sentence into words
words = sentence.split()

# Converting list to set to remove duplicates
unique_words = set(words)

# Printing results
print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)
```

```python
# Problem 2: Highest Salary from Employee Data

# Employee salary dictionary
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

# To Find employee with highest salary
highest_employee = max(employees, key=employees.get)

# Storing highest salary
highest_salary = employees[highest_employee]

# Printing result
print("Highest Salary:", highest_employee, "-", highest_salary)
```

```python
# Problem 3: Find Maximum and Minimum Values

# Given list
numbers = [45, 22, 89, 10, 66]

# Initializing max and min with first element
maximum = numbers[0]
minimum = numbers[0]

# Comparing each element
for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

# Printing result
print("List:", numbers)
print("Max:", maximum)
print("Min:", minimum)
```

```python
# Problem 4: Count Products Above a Price Threshold

# Given list of product prices
prices = [450, 1200, 899, 1500, 300]

# Initializing counter
product_count = 0

# Checking each price
for price in prices:
    if price > 1000:
        product_count += 1

# Printing result
print("Products above 1000:", product_count)
```

```python
# Problem 5: Calculate Attendance Percentage


# Given attendance list
attendance = ["P", "P", "A", "P", "P"]

# Count total days
total_days = len(attendance)

# Count present days
present_count = 0

for status in attendance:
    if status == "P":
        present_count += 1

# Calculate percentage
attendance_percentage = (present_count / total_days) * 100

# Print result
print("Attendance Percentage:", attendance_percentage)
```

```python
# Problem 6: Remove Duplicate Phone Numbers

# Given list of phone numbers
phone_numbers = [9876543210, 9123456789, 9876543210]

# Removing duplicates by using set
unique_numbers = set(phone_numbers)

# Printing result
print("Unique phone numbers:", unique_numbers)
```

```python
# Problem 7: Count Character Frequency

# Given string
text = "python"

# Creating an empty dictionary
char_count = {}

# Count each character
for char in text:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1

# Print result
print(char_count)
```

```python
# Problem 8: Convert List to Tuple

# Given list
numbers = [10, 20, 30]

# Converting list to tuple
tuple = tuple(numbers)

# Display result
print("Tuple:", tuple)
```

```python
# Problem 9: Check if a Key Exists in a Dictionary

# Taking a dictionary
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

# Key to check
key_to_check = "Ravi"

# Checking if key exists
if key_to_check in employees:
    print("Employee exists")
else:
    print("Employee does not exist")

```

```python
# Problem 10: Calculate Average Marks

# Take a list of marks
marks = [70, 75, 80, 65]

# Calculating total marks
total = sum(marks)

# Calculating average
average = total / len(marks)

# Printing result
print("Average Marks:", average)

```
