```python
# Problem 1: Smart Light Controller

class SmartLight:
    
    def __init__(self, name):
        self.name = name
        self.status = "OFF"   # Default state
    
    def turn_on(self):
        self.status = "ON"
    
    def turn_off(self):
        self.status = "OFF"
    
    def display_status(self):
        print(f"{self.name} is {self.status}")


# Sample Input
light = SmartLight("Bedroom Light")

# Action: ON
light.turn_on()

# Display Output
light.display_status()
```

```python
# Problem 2: Employee ID Card System

class EmployeeIDCard:
    
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department
    
    def display_id_card(self):
        print("----- Employee ID Card -----")
        print(f"Name       : {self.name}")
        print(f"Employee ID: {self.emp_id}")
        print(f"Department : {self.department}")
        print("-----------------------------")


# Sample Input
employee = EmployeeIDCard("Rahul", "EMP102", "IT")

# Display ID Card
employee.display_id_card()
```

```python
# Problem 3: Mobile Contact Record

class Contact:
    
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
    
    def save_contact(self):
        print("Contact Saved")
    
    def display_contact(self):
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")


# Sample Input
contact1 = Contact("Anita", "9876543210")

contact1.save_contact()
contact1.display_contact()
```

```python
# Problem 4: Product Price Tag Generator

class Product:
    
    def __init__(self, name, price):
        self.name = name
        self.price = price
    
    def print_price_tag(self):
        print(f"Product: {self.name}")
        print(f"Price: ₹{self.price}")


# Sample Input
product1 = Product("Headphones", 2499)

# Display Price Tag
product1.print_price_tag()
```

```python
# Problem 5: Movie Rating Display System

class Movie:
    
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating
    
    def display_movie_details(self):
        print("----- Movie Details -----")
        print(f"Movie Name: {self.name}")
        print(f"Rating    : {self.rating}")
        print("--------------------------")


# Sample Input
movie1 = Movie("Inception", 4.8)

# Display Movie Details
movie1.display_movie_details()
```

```python
# Problem 6: Delivery Address Manager

class DeliveryAddress:
    
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address
    
    def print_delivery_details(self):
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")


# Sample Input
delivery1 = DeliveryAddress("Suman", "Hyderabad")

# Display Delivery Details
delivery1.print_delivery_details()
```
